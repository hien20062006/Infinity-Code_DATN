```python
# =========================================================
# VIETNAM ELECTRONICS BIG DATASET GENERATOR
# 10,000+ ROWS
# WORLD BANK + ELECTRONICS + CUSTOMER DATA
# =========================================================

import requests
import pandas as pd
import numpy as np
import random

# =========================================================
# RANDOM SEED
# =========================================================

np.random.seed(42)
random.seed(42)

# =========================================================
# WORLD BANK INDICATORS
# =========================================================

indicators = {

    # ECONOMY
    "NY.GDP.MKTP.CD": "GDP",
    "NY.GDP.PCAP.CD": "GDP_Per_Capita",

    # ELECTRONICS / TECHNOLOGY
    "TX.VAL.TECH.MF.ZS": "HighTech_Exports",
    "TX.VAL.ICTG.ZS.UN": "ICT_Goods_Exports",

    # TRADE
    "BX.GSR.GNFS.CD": "Exports_Goods_Services",
    "NE.IMP.GNFS.CD": "Imports_Goods_Services",
    "NE.TRD.GNFS.ZS": "Trade_Percent_GDP",

    # INTERNET / DIGITAL
    "IT.NET.USER.ZS": "Internet_Users",
    "IT.CEL.SETS.P2": "Mobile_Subscriptions",
    "IT.NET.SECR.P6": "Secure_Internet_Servers",

    # POPULATION
    "SP.POP.TOTL": "Population",
    "SP.URB.TOTL.IN.ZS": "Urban_Population",

    # ECONOMIC CONDITIONS
    "FP.CPI.TOTL.ZG": "Inflation_Rate",

    # MANUFACTURING
    "NV.IND.MANF.CD": "Manufacturing_Value",

    # IMPORT TAX
    "TM.TAX.MRCH.SM.AR.ZS": "Import_Tariff"
}

# =========================================================
# COUNTRY
# =========================================================

country_code = "VNM"
country_name = "Vietnam"

# =========================================================
# VIETNAM CITIES
# =========================================================

cities = [
    "Ho Chi Minh City",
    "Hanoi",
    "Da Nang",
    "Can Tho",
    "Hai Phong",
    "Bien Hoa",
    "Nha Trang"
]

city_weights = [35, 30, 10, 8, 7, 5, 5]

# =========================================================
# PRODUCT CATEGORIES
# =========================================================

product_categories = [
    "Smartphone",
    "Laptop",
    "Tablet",
    "Gaming",
    "Smart TV",
    "Audio Devices",
    "Smart Watch",
    "Electronic Components",
    "Semiconductor",
    "Home Appliances"
]

product_weights = [25, 20, 10, 10, 8, 7, 5, 7, 5, 3]

# =========================================================
# BRANDS
# =========================================================

brand_mapping = {

    "Smartphone": ["Samsung", "Apple", "Xiaomi", "Oppo"],
    "Laptop": ["Dell", "HP", "Asus", "Lenovo"],
    "Tablet": ["Apple", "Samsung", "Xiaomi"],
    "Gaming": ["Asus", "MSI", "Acer"],
    "Smart TV": ["Samsung", "LG", "Sony"],
    "Audio Devices": ["Sony", "JBL", "Apple"],
    "Smart Watch": ["Apple", "Samsung"],
    "Electronic Components": ["Intel", "Foxconn"],
    "Semiconductor": ["Intel", "Nvidia"],
    "Home Appliances": ["LG", "Samsung"]

}

# =========================================================
# SUPPLIERS
# =========================================================

suppliers = [
    "Samsung Vietnam",
    "Intel Vietnam",
    "Foxconn Vietnam",
    "LG Vietnam",
    "FPT Retail",
    "Digiworld",
    "Viettel Store"
]

# =========================================================
# WAREHOUSES
# =========================================================

warehouses = [
    "WH-HCM-01",
    "WH-HN-02",
    "WH-DN-03"
]

# =========================================================
# PAYMENT METHODS
# =========================================================

payment_methods = [
    "Credit Card",
    "Bank Transfer",
    "E-Wallet",
    "Cash"
]

payment_weights = [30, 25, 35, 10]

# =========================================================
# SHIPPING METHODS
# =========================================================

shipping_methods = [
    "Air",
    "Road",
    "Sea"
]

shipping_weights = [30, 50, 20]

# =========================================================
# CUSTOMER SEGMENTS
# =========================================================

segments = [
    "Budget",
    "Standard",
    "Premium",
    "Enterprise"
]

segment_weights = [30, 40, 20, 10]

# =========================================================
# STORAGE
# =========================================================

all_data = []

# =========================================================
# FETCH WORLD BANK DATA
# =========================================================

print("Fetching Data From World Bank API...\n")

for indicator_code, indicator_name in indicators.items():

    url = f"https://api.worldbank.org/v2/country/{country_code}/indicator/{indicator_code}?format=json&per_page=100"

    response = requests.get(url)

    data = response.json()

    if len(data) > 1:

        records = data[1]

        for row in records:

            if row["value"] is not None:

                # ======================================
                # GENERATE MANY TRANSACTIONS
                # ======================================

                transaction_count = random.randint(25, 60)

                for _ in range(transaction_count):

                    # ======================================
                    # CITY
                    # ======================================

                    city = random.choices(
                        cities,
                        weights=city_weights,
                        k=1
                    )[0]

                    # ======================================
                    # REGION
                    # ======================================

                    if city in ["Hanoi", "Hai Phong"]:
                        region = "North"

                    elif city in ["Da Nang", "Nha Trang"]:
                        region = "Central"

                    else:
                        region = "South"

                    # ======================================
                    # PRODUCT CATEGORY
                    # ======================================

                    product = random.choices(
                        product_categories,
                        weights=product_weights,
                        k=1
                    )[0]

                    # ======================================
                    # BRAND
                    # ======================================

                    brand = random.choice(
                        brand_mapping[product]
                    )

                    # ======================================
                    # PRICE LOGIC
                    # ======================================

                    price_ranges = {

                        "Smartphone": (300, 1500),
                        "Laptop": (500, 2500),
                        "Tablet": (250, 1200),
                        "Gaming": (800, 3000),
                        "Smart TV": (400, 2000),
                        "Audio Devices": (50, 600),
                        "Smart Watch": (100, 900),
                        "Electronic Components": (20, 500),
                        "Semiconductor": (100, 3000),
                        "Home Appliances": (150, 2500)

                    }

                    low, high = price_ranges[product]

                    unit_price = round(
                        np.random.uniform(low, high),
                        2
                    )

                    quantity = np.random.randint(1, 20)

                    total_sales = round(
                        unit_price * quantity,
                        2
                    )

                    cost = round(
                        total_sales *
                        np.random.uniform(0.6, 0.85),
                        2
                    )

                    profit = round(
                        total_sales - cost,
                        2
                    )

                    # ======================================
                    # SHIPPING
                    # ======================================

                    shipping = random.choices(
                        shipping_methods,
                        weights=shipping_weights,
                        k=1
                    )[0]

                    if shipping == "Air":
                        delivery_days = np.random.randint(1, 4)

                    elif shipping == "Sea":
                        delivery_days = np.random.randint(10, 25)

                    else:
                        delivery_days = np.random.randint(3, 8)

                    # ======================================
                    # PURCHASE DATE
                    # ======================================

                    purchase_date = pd.Timestamp(
                        year=int(row["date"]),
                        month=random.randint(1, 12),
                        day=random.randint(1, 28)
                    )

                    # ======================================
                    # FINAL ROW
                    # ======================================

                    all_data.append({

                        # MACRO DATA
                        "Country": country_name,
                        "Indicator": indicator_name,
                        "Indicator_Code": indicator_code,
                        "Indicator_Value": row["value"],

                        # DATE
                        "Year": purchase_date.year,
                        "Month": purchase_date.month,
                        "Quarter": purchase_date.quarter,
                        "Purchase_Date": purchase_date,

                        # LOCATION
                        "City": city,
                        "Region": region,

                        # PRODUCT
                        "Product_Category": product,
                        "Brand": brand,

                        # CUSTOMER
                        "Gender": random.choice(["Male", "Female"]),
                        "Age": np.random.randint(18, 60),

                        "Customer_Segment": random.choices(
                            segments,
                            weights=segment_weights,
                            k=1
                        )[0],

                        # SALES
                        "Unit_Price_USD": unit_price,
                        "Quantity": quantity,
                        "Total_Sales_USD": total_sales,
                        "Cost_USD": cost,
                        "Profit_USD": profit,

                        # SHIPPING
                        "Shipping_Method": shipping,
                        "Delivery_Days": delivery_days,

                        # PAYMENT
                        "Payment_Method": random.choices(
                            payment_methods,
                            weights=payment_weights,
                            k=1
                        )[0],

                        # BUSINESS
                        "Import_Export": random.choice(
                            ["Import", "Export"]
                        ),

                        "Supplier": random.choice(
                            suppliers
                        ),

                        "Warehouse": random.choice(
                            warehouses
                        ),

                        # TAX
                        "Tax_Rate": random.choice(
                            [5, 8, 10]
                        ),

                        # CUSTOMER EXPERIENCE
                        "Satisfaction_Score": round(
                            np.random.uniform(3.0, 5.0),
                            1
                        )
                    })

# =========================================================
# CREATE DATAFRAME
# =========================================================

df = pd.DataFrame(all_data)

# =========================================================
# EXPORT FILES
# =========================================================

csv_file = "Vietnam_Electronics_BigDataset.csv"
excel_file = "Vietnam_Electronics_BigDataset.xlsx"

df.to_csv(csv_file, index=False)
df.to_excel(excel_file, index=False)

# =========================================================
# FINAL OUTPUT
# =========================================================

print("=" * 60)
print("DATASET CREATED SUCCESSFULLY")
print("=" * 60)

print("\nRows:", len(df))
print("Columns:", len(df.columns))

print("\nFILES CREATED:")
print("1.", csv_file)
print("2.", excel_file)

print("\nFINAL COLUMNS:\n")

for col in df.columns:
    print("-", col)

print("\nSAMPLE DATA:\n")
print(df.head())
```
